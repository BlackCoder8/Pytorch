### Matrix multiplication

Two main ways of performing multiplicaion in neural networks and deep learning:
1. Element-wise multiplication
2. Matrix multiplication (dot product)

THere are two main rules that performing matrix multiplication needs to satisfy:
1. The **inner dimentions** must match:

*`(3,2) @ (3,2)` wont work

*`(2,3) @ (3,2)` will work

*`(3,2) @ (2,3)` will work

2. The resulting matrix has the sape of the **outer dimentions**:

*(2, 3) @ (3, 2) -> (2, 2)

*(3, 2) @ (2, 3) -> (3, 3)

##One of the most common errors in deep learning is: shape errors

```python
#shapes for matrix multiplicatiosn
tensor_A=torch.tensor([[1,2],
                       [3,4],
                       [5,6]])

tensor_B=torch.tensor([[7,10],
                      [8, 11],
                      [9,12]])

torch.matmul(tensor_A, tensor_B)
```

```python
tensor_A.shape, tensor_B.shape
```

To fix our tensor sahpe issue, we can manupulate the shape of one of our tesnors using a **transpose**.

A **transpose** switches the axes or dimensions of a given tensor

```python
tensor_B, tensor_B.shape
```

```python
#with the help of transpose
tensor_B.T, tensor_B.T.shape
```

```python
#The matrix multiplication operaion works when tensor_B is transposed
print(f"Original shape: tensor_A = {tensor_A.shape}, tensor_B= {tensor_B.shape}")
print(f"New shape: tensor_A= {tensor_A.shape}, tensor_B.T= {tensor_B.T.shape}")
print(f"Multiplying: {tensor_A.shape} @ {tensor_B.T.shape} <-inner dimnesions must match ")
print("Output:\n")
output = torch.matmul(tensor_A,tensor_B.T)
print(output)
print(f"\n output shoae: {output.shape}")
```

##Reshaping,stacking, squeezing and unsqueezing tensor

* reshaping- reshapes an input tensor to a defeined shape

* View - return a view of an input tensor of cirtain shape but keep the same memory as the original tesor

* Stacking - combine multiple tensor on top of each other (vstack) or side by side (hsatck)

* Squeeze - rempve all `1` dimention to a target tensor

* Unsqueeze - add a `1` dimention to a target tensor

* permute - Return a view of the input with dimentions permuted (swapped) in a certain way

```python
import torch
```

```python
#Lets create a tensor

x = torch.arange(1.,10.)
x , x.shape
```

```python
#Add a extra dimention
x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape
```

```python
#Change the view
z = x.view(1, 9)
z, z.shape
```

```python
#changing z changes the x (because a view of a tensor shares the same memory as the original input)
z[:, 0] = 5
z, x
```

```python
#Stack tensor on top of each other
x_stacked = torch.stack([x,x,x,x] , dim=1)
x_stacked
```

```python
#torch.squeeze()- removes all single dimensions from a target tensor
print(f"Previous tesnor{x_reshaped}")
print(f"Previous shape {x_reshaped.shape}")

#remove extra dimentions from x_reshaped
x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
```

```python
#torch.unsqueeze() -  adds a single dimention to a  target tensor at a specific dim(dimentions)
print(f"Previous target: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

#Add an extra dimention with unsqueeze

x_unsqueezed = x_squeezed.unsqueeze(dim = 0)
print(f"\nNew tesnor: {x_unsqueezed}")
print(f"\nNew tensor: {x_unsqueezed.shape}")
```

```python
x_reshaped.squeeze()
```

```python
#torch.permute - rearrange the dimnetions of a target tensor in a specific order
x_original =torch.rand(size=(224, 224, 3)) #[height, width, color_chanels]

# permute the original tensor to rearrange the axis (or dim) orde
x_permuted = x_original.permute(2, 0, 1) #shifts 0->1, 1->2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")#[color_chanel, height, width]
```
